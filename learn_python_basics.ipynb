## 1. Variables
A variable is just a name that holds a value. Think of it like a labeled box.

```python
# The # symbol means this line is a comment — Python ignores it
# Comments are notes for humans reading the code

age = 39
name = "Hansel"
salary_over_50k = False

print(age)
print(name)
print(salary_over_50k)
```

```python
# You can combine variables in a print statement using f-strings
# f-strings let you put variables directly inside text

print(f"{name} is {age} years old.")
```

## 2. Data Types
Every value in Python has a type. The main ones are:

```python
integer_number = 39          # int   — whole number
decimal_number = 3.14        # float — number with decimal
text = "Hello"               # str   — text (string)
true_or_false = True         # bool  — True or False only

# type() tells you what type a variable is
print(type(integer_number))
print(type(decimal_number))
print(type(text))
print(type(true_or_false))
```

---
## 3. Lists
A list holds multiple values in one variable. Your dataset has columns — each column is basically a list.

```python
# A list of job types from your dataset
job_types = ["State-gov", "Self-emp-not-inc", "Private", "Federal-gov"]

print(job_types)
print(f"Number of job types: {len(job_types)}")
```

```python
# Access items by position — positions start at 0, not 1!
print(job_types[0])   # first item
print(job_types[1])   # second item
print(job_types[-1])  # last item
```

```python
# Add something to a list
job_types.append("Without-pay")
print(job_types)
```

---
## 4. Loops
A loop lets you repeat code for each item in a list — instead of writing the same line over and over.

```python
# Without a loop — repetitive and bad practice
print(job_types[0])
print(job_types[1])
print(job_types[2])
```

```python
# With a loop — works for any size list
for job in job_types:
    print(job)
```

```python
# You can also loop over a range of numbers
for i in range(5):
    print(f"Row number: {i}")
```

---
## 5. If / Else — Making Decisions
Code can make decisions based on conditions.

```python
salary_range = ">50K"

if salary_range == ">50K":
    print("This person earns more than $50,000")
else:
    print("This person earns $50,000 or less")
```

```python
age = 25

if age < 18:
    print("Under 18")
elif age < 30:
    print("Young adult")
elif age < 60:
    print("Working age")
else:
    print("Senior")
```

---
## 6. Functions
A function is reusable code. Instead of writing the same logic multiple times, you define it once and call it.

```python
# Define a function
def describe_person(age, job, salary):
    print(f"Age: {age}, Job: {job}, Salary range: {salary}")

# Call the function
describe_person(39, "State-gov", "<=50K")
describe_person(50, "Self-emp-not-inc", "<=50K")
describe_person(52, "Self-emp-inc", ">50K")
```

```python
# Functions can also return a value
def age_group(age):
    if age < 30:
        return "Young"
    elif age < 50:
        return "Middle-aged"
    else:
        return "Senior"

print(age_group(25))
print(age_group(39))
print(age_group(65))
```

## 7. Connecting to Your Real Data
Now let's apply what you learned to your actual dataset.

```python
import pandas as pd

# Load the data
df = pd.read_csv("data/adultdata.csv")

# df is a DataFrame — think of it as a spreadsheet in Python
print(f"The dataset has {df.shape[0]} rows and {df.shape[1]} columns")
```

```python
# Look at the first 5 rows
df.head()
```

```python
# Each column name is a string — let's loop through them
for column in df.columns:
    print(column)
```

```python
# Use the age_group function we wrote on real data
df['age_group'] = df['age'].apply(age_group)

# Count how many people are in each group
df['age_group'].value_counts()
```
